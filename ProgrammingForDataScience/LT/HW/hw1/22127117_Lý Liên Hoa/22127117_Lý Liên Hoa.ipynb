## Import libraries

```python
import time
import numpy as np
import matplotlib.pyplot as plt
```

## Using basic Python to multiply 2 matrices n x n

```python
def python_multiply_matrix(n):
    a = np.random.rand(n, n)
    b = np.random.rand(n, n)
    result = np.zeros((n, n))

    # Start time
    start_time = time.time()
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                result[i][j] += a[i][k] + b[k][j]

    # End time
    end_time = time.time()

    return end_time - start_time
```

## Using Numpy to multiply 2 matrices n x n

```python
def numpy_multiply_matrices(n):
    a = np.random.rand(n, n)
    b = np.random.rand(n, n)
    
    start_time = time.time()
    result = np.dot(a, b)
    end_time = time.time()
    return end_time - start_time
```

## Testing

```python
# sizes = [1, 100, 500, 1000, 2500, 3500, 4500, 5000]
sizes = [1, 100, 500, 1000, 2500]
sizes
```

### Result of using basic Python

```python
python_times = []

for n in sizes:
    python_times.append(python_multiply_matrix(n))

python_times
```

### Result of using Numpy

```python
numpy_times = []

for n in sizes:
    numpy_times.append(numpy_multiply_matrices(n))
    
numpy_times
```

### Result of using C
The result obtained after the program C was compiled

```python
# Runtime of matrices with size 1, 100, 500, 1000, 2500, 3500, 4500, 5000 when using C
# c_times = [0, 0.002, 0.709, 9.423, 200.343, 574.41, 1162.285, 1649.548]
c_times = [0, 0.002, 0.709, 9.423, 200.343]
c_times
```

## Visualization

```python
plt.figure()

# Vẽ thời gian chạy của Python
plt.plot(sizes, python_times, label='Python', marker='o')

# Vẽ thời gian chạy của NumPy
plt.plot(sizes, numpy_times, label='NumPy', marker='s')

# Vẽ thời gian chạy của C
plt.plot(sizes, c_times, label='C', marker='x', linestyle='-')

# Đặt thang đo log-log
plt.xscale('log')
plt.yscale('log')

# Thêm nhãn trục và tiêu đề
plt.xlabel('Kích thước ma trận (n)')
plt.ylabel('Thời gian (giây)')
plt.title('So sánh thời gian chạy của các phương pháp nhân ma trận')

# Hiển thị chú giải biểu đồ
plt.legend()

plt.show()
```

## Conclusion

From the line chart, we can conclude that:
- Using basic python has the lowest performance when performing large matrix multiplication.
- Numpy and C both are better choice for performing large matrix multiplication.
- It can be said that coding in C language is the optimal choice.

Explanation:
- Python is a high-level language so that Python cannot access hardware directly which can lead to slow down the execution time.
- Although written in Python, Numpy uses low-level libraries written in C, so it can perform the mathematical operation faster.
- Finally, here, we use same algorithm to multiply 2 matrices in Python and C but C's execution time is much better. C is a low-level language, and it provides direct access to hardware so that C code compiled to machine code very fast.
