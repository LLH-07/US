## **Tính xác suất p của biến cố có ít nhất 3 người cùng sinh nhật (cùng ngày và tháng sinh) trong nhóm k người?**


### **Xác suất không có ai cùng ngày sinh**
$$
p_0 = P(A) = 1 - P(B) = 1 - \frac{P_{365}^k}{365^k} = 1 - \frac{365!}{(365 - k)!365^k}.
$$

### **Xác suất có chính xác 2 người cùng ngày sinh**
Chọn ngẫu nhiên 2 người trong nhóm k người: 
$$
C_{k}^{2}
$$

Số cách chọn ngày sinh cho người thứ nhất được chọn: 
$$
365
$$

Số cách chọn ngày sinh cho người thứ hai được chọn: 
$$
1
$$

Số cách chọn ngày sinh cho **(k - 2)** người còn lại:
$$
{P_{364}^{k- 2}} = \frac{364!}{(364 - (k-2))!}
$$

#### Xác suất có chính xác 2 người cùng ngày sinh là:
$$
p_1 = \frac{C_{k}^{2} . 365 . 1 . {P_{364}^{k- 2}}}{365^k}
$$

### **Xác suất có ít nhất 3 người cùng ngày sinh trong nhóm k người**
$$
p = 1 - p_0 - p_1
$$

```python
import numpy as np
import math
```

```python

# Xác suất không có ai có cùng ngày sinh
def birthday_probability_no_same(k):
    if k > 365: 
        return 0
    
    p = 1.0
    for i in range(k):
        p *= (365 - i) / 365
        
    return p

def birthday_probability_exact_2(k):
    if k < 2 or k > 365:
        return 0
    
    p = 1.0
    
    # Lấy tổ 2 người cùng ngày sinh nhật trong số k người
    comb_k_2 = math.comb(k, 2)
    
    p_2 = 1 / 365 # (365 * 1 / 365 ^ 2)
    
    p_rest = 1.0
    
    for i in range(k - 2):
        p_rest *= (364 - i) / 365
        
    result = comb_k_2 * p_2 * p_rest
    
    return result
    
def birthday_probability_at_least_3(k):
    if k < 3:
        return 0
    
    return 1 - birthday_probability_no_same(k) - birthday_probability_exact_2(k)
    
    
    
```

```python
birthday_probability_at_least_3(30)
```
