```python
colors1 = plt.get_cmap('tab20c', 20)
colors2 = plt.get_cmap('tab20', 20)

colors = np.vstack((colors1.colors, colors2.colors))

idx = np.arange(40)
np.random.shuffle(idx)

colors = colors[idx]
```

```python
def plot_gdp_by_year(data, ds_quocgia, title):

    plt.figure(figsize=(15, 10))

    plt.title(title)

    for i, qg in enumerate(ds_quocgia):
        qg_df = data[data['Quốc gia'] == qg]
        plt.plot(qg_df['Năm'], qg_df['GDP'], label=qg, marker='o', color=colors[i])

    plt.xlabel('Năm')
    plt.ylabel('GDP')

    plt.xticks(df_simplified['Năm'].unique())

    ticks = plt.gca().get_yticks()
    plt.gca().set_yticks(ticks)
    plt.gca().set_yticklabels([f'{int(tick * 1e-9):.1f}B' for tick in ticks])

    plt.ylim(bottom=0)

    plt.grid(True, alpha=0.5)
    plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1), title='Quốc gia')

    plt.show()
```

### **2. Tác động của xuất khẩu và nhập khẩu đến GDP các quốc gia như thế nào?**

Để phân tích ta sẽ chọn ra 5 quốc gia có GDP trung bình cao nhất và thấp nhất để so sánh, phân tích ảnh hưởng của xuất nhập khẩu ảnh hưởng như thế nào tới GDP của quốc gia đó

```python
top_5_lowest_df = df_simplified.groupby('Quốc gia')['GDP'].mean().sort_values(ascending=True).head(5)

top_5_highest_df = df_simplified.groupby('Quốc gia')['GDP'].mean().sort_values(ascending=False).head(5)
```

#### **Xuất khẩu**

```python
def plot_line_exp(data, countries_lists, title):
    plt.figure(figsize=(15, 10))

    for i, country in enumerate(countries_lists):
        country_data = data[data["Quốc gia"] == country]
        plt.plot(country_data["Năm"], country_data["GDP"], label=f"{country} GDP", color=colors[i])
        plt.plot(country_data["Năm"], country_data["Xuất khẩu hàng hóa"], linestyle="--", label=f"{country} Xuất khẩu", color=colors[i])

    plt.xlabel("Năm")
    plt.ylabel("Tỷ USD")
    plt.title(title)

    plt.xticks(df_simplified['Năm'].unique())

    ticks = plt.gca().get_yticks()
    plt.gca().set_yticks(ticks)
    plt.gca().set_yticklabels([f'{int(tick * 1e-9):.1f}B' for tick in ticks])

    plt.ylim(bottom=0)

    plt.grid(True, alpha=0.5)

    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1), title='Quốc gia')
    plt.show()
```

```python
def plot_line_imp(data, countries_lists, title):
    plt.figure(figsize=(15, 10))

    for i, country in enumerate(countries_lists):
        country_data = data[data["Quốc gia"] == country]
        plt.plot(country_data["Năm"], country_data["GDP"], label=f"{country} GDP", color=colors[i])
        plt.plot(country_data["Năm"], country_data["Nhập khẩu hàng hóa"], linestyle="--", label=f"{country} Xuất khẩu", color=colors[i])

    plt.xlabel("Năm")
    plt.ylabel("Tỷ USD")
    plt.title(title)

    plt.xticks(df_simplified['Năm'].unique())

    ticks = plt.gca().get_yticks()
    plt.gca().set_yticks(ticks)
    plt.gca().set_yticklabels([f'{int(tick * 1e-9):.1f}B' for tick in ticks])

    plt.ylim(bottom=0)

    plt.grid(True, alpha=0.5)

    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1), title='Quốc gia')
    plt.show()
```

### **3. Một quốc gia đông dân, đặc biệt dân số ở độ tuổi lao động (15-64 tuổi) đông có GDP như nào? (2023)**

```python
df_2023 = df_cleaned[df_cleaned['Năm'] == 2023].drop(['Năm'], axis=1).reset_index(drop=True)

df_2023.head()
```

```python
working_age_female = [f'Dân số độ tuổi {i}-{i+4}, nữ (% dân số nữ)' for i in range(15, 65, 5)]
working_age_male = [f'Dân số độ tuổi {i}-{i+4}, nam (% dân số nam)' for i in range(15, 65, 5)]
```

```python
total_working_female_data = df_2023[working_age_female].sum(axis=1)
total_working_male_data = df_2023[working_age_male].sum(axis=1)

df_2023['Tổng lao động nữ'] = df_2023['Dân số nữ'] * total_working_female_data / 100
df_2023['Tổng lao động nam'] = df_2023['Dân số nam'] * total_working_male_data / 100


df_2023['Tổng lao động nữ (%)'] = df_2023['Tổng lao động nữ'] / df_2023['Tổng dân số'] * 100
df_2023['Tổng lao động nam (%)'] = df_2023['Tổng lao động nam'] / df_2023['Tổng dân số'] * 100

df_2023['Tổng lao động'] = df_2023['Tổng lao động nữ'] + df_2023['Tổng lao động nam']


df_2023 = df_2023[['Quốc gia', 'Tổng dân số', 'Dân số nam', 'Dân số nữ', 'Tổng lao động nữ (%)', 'Tổng lao động nam (%)', 'Tổng lao động', 'Tỷ lệ tham gia lực lượng lao động (% tổng dân số từ 15-64 tuổi)', 'GDP']]
```

```python
df_2023.head()
```

```python
plt.figure(figsize=(12,8))
sns.scatterplot(data=df_2023, x='Tổng dân số', y='GDP', hue='Quốc gia', palette=colors, s=100)

plt.title('GDP và Tổng dân số (2023)')
plt.xlabel('Tổng dân số')
plt.ylabel('GDP (tỷ đôla Mỹ)')

ticks = plt.gca().get_yticks()
plt.gca().set_yticks(ticks)
plt.gca().set_yticklabels([f'{int(tick * 1e-9):.1f}B' for tick in ticks])

xticks = plt.gca().get_xticks()
plt.gca().set_xticks(xticks)
plt.gca().set_xticklabels([f'{int(tick * 1e-6):.1f}M' for tick in xticks]) 

plt.ylim(bottom=0)

plt.grid(True, alpha=0.5)
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1), title='Quốc gia')

plt.show()
```

Nhận xét:
- Mối tương quan giữa tổng dân số và GDP không quá lớn. 
- Các quốc gia có dân số lớn không nhất thiết sẽ có GDP cao và ngược lại. Ví dụ, Indonesia có tổng dân số khoảng 280 triệu người nhưng lại có GDP thấp hơn Nhật Bản với tổng dân số khoảng 130 triệu người.
